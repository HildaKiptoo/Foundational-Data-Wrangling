# 1. Creating a DataFrame from a NumPy Array

# 3. Imports & Data Generation

```python
# Libraries needed: pandas, numpy
import pandas as pd
import numpy as np

# Setting a seed so the random data is the same every time the cell runs
np.random.seed(42)

# Generating 1,000 rows of synthetic student performance data
num_students = 1000
data = {
    'gender': np.random.choice(['female', 'male'], num_students),
    'lunch_type': np.random.choice(['standard', 'free/reduced'], num_students, p=[0.65, 0.35]),
    'test_prep': np.random.choice(['none', 'completed'], num_students),
    'math_score': np.random.normal(65, 15, num_students).clip(0, 100).astype(int),
    'reading_score': np.random.normal(69, 14, num_students).clip(0, 100).astype(int),
}

df_exams = pd.DataFrame(data)

# Injecting some realistic "messy" missing data to teach handling nulls
df_exams.loc[15:25, 'math_score'] = np.nan

# Displaying the first 5 rows
df_exams.head()
```

# 4. Inspecting the DataFrame

```python
# Libraries needed: pandas
# Checking the dimensions (rows, columns)
print(f"Dataset shape: {df_exams.shape}\n")

# Getting a concise summary of columns, non-null counts, and datatypes
df_exams.info()

# Generating descriptive statistics for the numerical columns
df_exams.describe().round(2)
```

# 5. Foundational Data Wrangling

```python
# Libraries needed: pandas
# 1. Handling Missing Values
# Filling the nulls we injected earlier with the column's mean score
mean_math = df_exams['math_score'].mean()
df_exams['math_score'] = df_exams['math_score'].fillna(mean_math).astype(int)

# 2. Creating Calculated Columns
df_exams['total_score'] = df_exams['math_score'] + df_exams['reading_score']

# 3. Filtering Data
# Finding students who completed prep and scored highly
top_performers = df_exams[(df_exams['test_prep'] == 'completed') & (df_exams['total_score'] >= 170)]

print(f"Number of top performers: {top_performers.shape[0]}")
top_performers.head()
```

# 6. Grouping & Aggregation

```python
# Libraries needed: pandas
# Analyzing average scores based on whether students completed the prep course
prep_summary = df_exams.groupby('test_prep')[['math_score', 'reading_score', 'total_score']].mean()

# Sorting the results for cleaner presentation
prep_summary.sort_values(by='total_score', ascending=False).round(2)
```

# 7. Counting Categorical Data (Value Counts)

```python
# Libraries needed: pandas
# Checking the raw count of students in each lunch category
lunch_counts = df_exams['lunch_type'].value_counts()

# Showing the percentage breakdown (normalize=True)
lunch_percentages = df_exams['lunch_type'].value_counts(normalize=True).round(4) * 100

print("Count of students by lunch type:")
print(lunch_counts)
print("\nPercentage breakdown:")
print(lunch_percentages)
```

# 8. Conditional Logic (Creating a Pass/Fail Column)

```python
# Libraries needed: pandas, numpy
# Creating a new column 'status' based on the total score
pass_mark = 130
df_exams['status'] = np.where(df_exams['total_score'] >= pass_mark, 'Pass', 'Fail')

# Checking how many students passed vs failed
df_exams['status'].value_counts()
```

# 9. Introduction to Pivot Tables

```python
# Libraries needed: pandas
# Creating a pivot table showing the average reading score
# breaking down by gender (rows) and test preparation (columns)
pivot_summary = pd.pivot_table(
    df_exams,
    values='reading_score',
    index='gender',
    columns='test_prep',
    aggfunc='mean'
)

pivot_summary.round(2)
```

# 10. Built-in Pandas Plotting

```python
# Libraries needed: matplotlib, pandas
import matplotlib.pyplot as plt

# Plotting a histogram to show the distribution of math scores
histogram = df_exams['math_score'].plot(
    kind='hist',
    bins=20,
    title='Distribution of Math Scores',
    color='skyblue',
    edgecolor='black',
    figsize=(8, 5)
)
```

# 11. Seaborn Box Plot

```python
# Libraries needed: matplotlib, seaborn
import seaborn as sns
import matplotlib.pyplot as plt

# Setting a clean visual theme
sns.set_theme(style="whitegrid")

# Initializing the figure size (width, height in inches)
plt.figure(figsize=(10, 6))

# Creating a box plot comparing math scores across genders, splitting by test prep
sns.boxplot(
    data=df_exams,
    x='gender',
    y='math_score',
    hue='test_prep',
    palette='Set2'
)

# Adding descriptive titles and adjusting font sizes
plt.title('Math Scores by Gender and Test Preparation', fontsize=14)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Math Score', fontsize=12)

# Moving the legend outside the main plot area so it doesn't overlap data
plt.legend(title='Test Prep', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjusting layout to prevent clipping and displaying the static chart
plt.tight_layout()
plt.show()
```

# 12. Matplotlib Pie Chart

```python
# Libraries needed: matplotlib
import matplotlib.pyplot as plt

# Counting the occurrences of each status
status_counts = df_exams['status'].value_counts()

# Setting up the figure size
plt.figure(figsize=(7, 7))

# Creating the pie chart
plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['#66b3ff', '#ff9999'],
    explode=(0.05, 0),
    shadow=True
)
plt.title('Overall Pass vs. Fail Distribution', fontsize=14, fontweight='bold')

# Ensuring the pie is drawn as a circle
plt.axis('equal')
plt.show()
```

# 13. Plotly Interactive Scatter Plot

```python
# Libraries needed: plotly
import plotly.express as px

# Create an interactive scatter plot
fig = px.scatter(
    df_exams,
    x='math_score',
    y='reading_score',
    color='test_prep',
    title='Math vs. Reading Scores by Test Preparation',
    labels={
        'math_score': 'Math Score',
        'reading_score': 'Reading Score',
        'test_prep': 'Test Prep Status'
    },
    opacity=0.7
)

# Display the interactive plot
fig.show()
```
